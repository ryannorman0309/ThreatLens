```python
# ============================================================
# THREATLENS - TASK 7
# STEP 1: VERIFY PREPARED DATA
# ============================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("THREATLENS - TASK 7")
print("STEP 1: VERIFYING PREPARED DATA")
print("=" * 70)


# ============================================================
# Data Loading (Added to resolve NameError)
# ============================================================
# Assuming the CSV files are available in the current directory
X_train_final = pd.read_csv('/content/ThreatLens_X_train.csv')
X_test_final = pd.read_csv('/content/ThreatLens_X_test.csv')
y_train_encoded = pd.read_csv('/content/ThreatLens_y_train.csv').values.flatten()
y_test_encoded = pd.read_csv('/content/ThreatLens_y_test.csv').values.flatten()


# ============================================================
# STEP 1.1 - CHECK SHAPES
# ============================================================

print("\n[1.1] Dataset shapes")

print("X_train_final :", X_train_final.shape)
print("X_test_final  :", X_test_final.shape)

print("y_train_encoded :", y_train_encoded.shape)
print("y_test_encoded  :", y_test_encoded.shape)


# ============================================================
# STEP 1.2 - CHECK DATA TYPES
# ============================================================

print("\n[1.2] Data types")

print("X_train dtype:",
      X_train_final.dtype if hasattr(X_train_final, "dtype")
      else "DataFrame")

print("X_test dtype:",
      X_test_final.dtype if hasattr(X_test_final, "dtype")
      else "DataFrame")


# ============================================================
# STEP 1.3 - CHECK FOR MISSING VALUES
# ============================================================

print("\n[1.3] Checking missing values")

if isinstance(X_train_final, pd.DataFrame):

    print(
        "Training missing values:",
        X_train_final.isnull().sum().sum()
    )

    print(
        "Testing missing values:",
        X_test_final.isnull().sum().sum()
    )

else:

    print(
        "Training missing values:",
        np.isnan(X_train_final).sum()
    )

    print(
        "Testing missing values:",
        np.isnan(X_test_final).sum()
    )


# ============================================================
# STEP 1.4 - CHECK TARGET CLASSES
# ============================================================

print("\n[1.4] Target classes")

unique_train = np.unique(y_train_encoded)
unique_test = np.unique(y_test_encoded)

print("Training classes:", unique_train)
print("Testing classes :", unique_test)

print(
    "\nNumber of classes:",
    len(unique_train)
)


# ============================================================
# STEP 1.5 - CLASS DISTRIBUTION
# ============================================================

print("\n[1.5] Training class distribution")

train_distribution = pd.Series(
    y_train_encoded
).value_counts().sort_index()

display(train_distribution)


print("\n[1.6] Testing class distribution")

test_distribution = pd.Series(
    y_test_encoded
).value_counts().sort_index()

display(test_distribution)


# ============================================================
# FINAL CHECK
# ============================================================

print("\n" + "=" * 70)
print("STEP 1 COMPLETED")
print("=" * 70)

print("""
✓ Training features verified
✓ Testing features verified
✓ Target labels verified
✓ Missing values checked
✓ Number of classes identified
✓ Class distribution checked

Next:
Step 2 → Prepare data for LSTM
""")
```

```python
# ============================================================
# THREATLENS - TASK 7
# STEP 2: PREPARE DATA FOR LSTM
# ============================================================

print("=" * 70)
print("THREATLENS - TASK 7")
print("STEP 2: PREPARING DATA FOR LSTM")
print("=" * 70)


# ============================================================
# STEP 2.1 - CONVERT FEATURES TO NUMPY
# ============================================================

print("\n[2.1] Converting feature matrices...")


if isinstance(X_train_final, pd.DataFrame):

    X_train_lstm = X_train_final.to_numpy(
        dtype=np.float32
    )

else:

    X_train_lstm = np.asarray(
        X_train_final,
        dtype=np.float32
    )


if isinstance(X_test_final, pd.DataFrame):

    X_test_lstm = X_test_final.to_numpy(
        dtype=np.float32
    )

else:

    X_test_lstm = np.asarray(
        X_test_final,
        dtype=np.float32
    )


# ============================================================
# STEP 2.2 - CONVERT TARGETS
# ============================================================

y_train_lstm = np.asarray(
    y_train_encoded,
    dtype=np.int32
)

y_test_lstm = np.asarray(
    y_test_encoded,
    dtype=np.int32
)


# ============================================================
# STEP 2.3 - DISPLAY SHAPES
# ============================================================

print("\nTraining feature matrix:",
      X_train_lstm.shape)

print("Testing feature matrix:",
      X_test_lstm.shape)

print("Training labels:",
      y_train_lstm.shape)

print("Testing labels:",
      y_test_lstm.shape)


# ============================================================
# STEP 2.4 - CHECK FINITE VALUES
# ============================================================

print("\n[2.4] Checking numerical values...")

print(
    "Training NaN:",
    np.isnan(X_train_lstm).sum()
)

print(
    "Testing NaN:",
    np.isnan(X_test_lstm).sum()
)

print(
    "Training infinite:",
    np.isinf(X_train_lstm).sum()
)

print(
    "Testing infinite:",
    np.isinf(X_test_lstm).sum()
)


# ============================================================
# FINAL
# ============================================================

print("\n" + "=" * 70)
print("STEP 2 COMPLETED")
print("=" * 70)

print("""
✓ Feature matrices converted to NumPy
✓ Features converted to float32
✓ Target labels converted to integer format
✓ Numerical values verified

Next:
Step 3 → Convert data into LSTM sequence format
""")
```

```python
# ============================================================
# THREATLENS - TASK 7
# STEP 3: CREATE LSTM SEQUENCE FORMAT
# ============================================================

print("=" * 70)
print("THREATLENS - TASK 7")
print("STEP 3: CREATING LSTM SEQUENCE FORMAT")
print("=" * 70)


# ============================================================
# STEP 3.1 - ORIGINAL SHAPE
# ============================================================

print("\nOriginal training shape:")
print(X_train_lstm.shape)

print("\nOriginal testing shape:")
print(X_test_lstm.shape)


# ============================================================
# STEP 3.2 - ADD TIMESTEP DIMENSION
# ============================================================

X_train_lstm = X_train_lstm.reshape(
    X_train_lstm.shape[0],
    1,
    X_train_lstm.shape[1]
)

X_test_lstm = X_test_lstm.reshape(
    X_test_lstm.shape[0],
    1,
    X_test_lstm.shape[1]
)


# ============================================================
# STEP 3.3 - DISPLAY FINAL SHAPE
# ============================================================

print("\nFinal LSTM training shape:")
print(X_train_lstm.shape)

print("\nFinal LSTM testing shape:")
print(X_test_lstm.shape)


# ============================================================
# EXTRACT DIMENSIONS
# ============================================================

num_training_samples = X_train_lstm.shape[0]
num_testing_samples = X_test_lstm.shape[0]

timesteps = X_train_lstm.shape[1]

num_features = X_train_lstm.shape[2]

num_classes = len(
    np.unique(y_train_lstm)
)


print("\nLSTM Input Information")
print("-" * 50)

print(
    "Training samples:",
    num_training_samples
)

print(
    "Testing samples:",
    num_testing_samples
)

print(
    "Timesteps:",
    timesteps
)

print(
    "Features per timestep:",
    num_features
)

print(
    "Output classes:",
    num_classes
)


# ============================================================
# FINAL
# ============================================================

print("\n" + "=" * 70)
print("STEP 3 COMPLETED")
print("=" * 70)

print("""
LSTM input format:

(samples, timesteps, features)

The current ThreatLens data is represented as:

(samples, 1, features)

Next:
Step 4 → Build the LSTM architecture
""")
```

```python
# ============================================================
# THREATLENS - TASK 7
# STEP 4: BUILD LSTM ARCHITECTURE
# ============================================================

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    LSTM,
    Dense,
    Dropout
)

print("=" * 70)
print("THREATLENS - TASK 7")
print("STEP 4: BUILDING LSTM ARCHITECTURE")
print("=" * 70)


# ============================================================
# STEP 4.1 - SET RANDOM SEEDS
# ============================================================

tf.random.set_seed(42)
np.random.seed(42)


# ============================================================
# STEP 4.2 - CREATE MODEL
# ============================================================

lstm_model = Sequential([

    LSTM(
        64,
        input_shape=(
            timesteps,
            num_features
        )
    ),

    Dropout(0.30),

    Dense(
        32,
        activation="relu"
    ),

    Dropout(0.20),

    Dense(
        num_classes,
        activation="softmax"
    )
])


# ============================================================
# STEP 4.3 - DISPLAY ARCHITECTURE
# ============================================================

print("\nLSTM Architecture:")
print("-" * 70)

lstm_model.summary()


# ============================================================
# FINAL
# ============================================================

print("\n" + "=" * 70)
print("STEP 4 COMPLETED")
print("=" * 70)

print("""
Architecture:

Input
  ↓
LSTM - 64 units
  ↓
Dropout - 30%
  ↓
Dense - 32 neurons
  ↓
Dropout - 20%
  ↓
Softmax Output

Next:
Step 5 → Configure optimizer, loss function and hyperparameters
""")
```

```python
# ============================================================
# THREATLENS - TASK 7
# STEP 5: COMPILE LSTM MODEL
# ============================================================

print("=" * 70)
print("THREATLENS - TASK 7")
print("STEP 5: COMPILING LSTM MODEL")
print("=" * 70)


# ============================================================
# STEP 5.1 - COMPILE
# ============================================================

lstm_model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="sparse_categorical_crossentropy",

    metrics=["accuracy"]
)


# ============================================================
# STEP 5.2 - HYPERPARAMETERS
# ============================================================

batch_size = 64

epochs = 30

validation_split = 0.20


print("\nHyperparameters")
print("-" * 50)

print(
    "Optimizer:",
    "Adam"
)

print(
    "Learning rate:",
    0.001
)

print(
    "Loss function:",
    "Sparse Categorical Crossentropy"
)

print(
    "Batch size:",
    batch_size
)

print(
    "Maximum epochs:",
    epochs
)

print(
    "Validation split:",
    validation_split
)


print("\n✓ LSTM model compiled successfully")


print("\n" + "=" * 70)
print("STEP 5 COMPLETED")
print("=" * 70)

print("""
Next:
Step 6 → Add regularization and training callbacks
""")
```

```python
# ============================================================
# THREATLENS - TASK 7
# STEP 6: REGULARIZATION AND CALLBACKS
# ============================================================

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)

print("=" * 70)
print("THREATLENS - TASK 7")
print("STEP 6: REGULARIZATION AND CALLBACKS")
print("=" * 70)


# ============================================================
# STEP 6.1 - EARLY STOPPING
# ============================================================

early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=5,

    restore_best_weights=True,

    verbose=1
)


# ============================================================
# STEP 6.2 - LEARNING RATE REDUCTION
# ============================================================

reduce_lr = ReduceLROnPlateau(

    monitor="val_loss",

    factor=0.5,

    patience=2,

    min_lr=0.00001,

    verbose=1
)


# ============================================================
# REGULARIZATION SUMMARY
# ============================================================

print("\nRegularization techniques")
print("-" * 50)

print("✓ Dropout after LSTM      : 30%")
print("✓ Dropout after Dense     : 20%")
print("✓ Early Stopping          : patience = 5")
print("✓ Learning Rate Reduction : factor = 0.5")


print("\n" + "=" * 70)
print("STEP 6 COMPLETED")
print("=" * 70)

print("""
Regularization and training callbacks are ready.

Next:
Step 7 → Train the LSTM model
""")
```

```python
# ============================================================
# THREATLENS - TASK 7
# STEP 7: TRAIN LSTM MODEL
# ============================================================

print("=" * 70)
print("THREATLENS - TASK 7")
print("STEP 7: TRAINING LSTM MODEL")
print("=" * 70)


# ============================================================
# STEP 7.1 - TRAIN MODEL
# ============================================================

print("\nStarting LSTM training...\n")


history = lstm_model.fit(

    X_train_lstm,

    y_train_lstm,

    epochs=epochs,

    batch_size=batch_size,

    validation_split=validation_split,

    callbacks=[
        early_stopping,
        reduce_lr
    ],

    verbose=1
)


# ============================================================
# STEP 7.2 - TRAINING SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("LSTM TRAINING COMPLETED")
print("=" * 70)

print(
    "\nEpochs actually completed:",
    len(history.history["loss"])
)

print(
    "Best validation accuracy:",
    f"{max(history.history['val_accuracy']):.4f}"
)

print(
    "Best validation loss:",
    f"{min(history.history['val_loss']):.4f}"
)


print("\n✓ LSTM model successfully trained")

print("\nNext:")
print("Step 8 → Visualize training and validation performance")
```

```python
# ============================================================
# THREATLENS - TASK 7
# STEP 8: TRAINING PERFORMANCE
# ============================================================

import matplotlib.pyplot as plt

print("=" * 70)
print("THREATLENS - TASK 7")
print("STEP 8: TRAINING PERFORMANCE")
print("=" * 70)


# ============================================================
# STEP 8.1 - ACCURACY
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title(
    "ThreatLens - LSTM Training and Validation Accuracy"
)

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.show()


# ============================================================
# STEP 8.2 - LOSS
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title(
    "ThreatLens - LSTM Training and Validation Loss"
)

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.show()


print("\n✓ Accuracy graph generated")
print("✓ Loss graph generated")


print("\n" + "=" * 70)
print("STEP 8 COMPLETED")
print("=" * 70)

print("""
The graphs can be used to analyze:

- Learning progression
- Training vs validation performance
- Possible overfitting
- Convergence of the model

Next:
Step 9 → Evaluate the trained LSTM on the test dataset
""")
```

```python
# ============================================================
# THREATLENS - TASK 7
# STEP 9: LSTM MODEL EVALUATION
# ============================================================

print("=" * 70)
print("THREATLENS - TASK 7")
print("STEP 9: LSTM MODEL EVALUATION")
print("=" * 70)


# ============================================================
# STEP 9.1 - TEST EVALUATION
# ============================================================

test_loss, test_accuracy = lstm_model.evaluate(

    X_test_lstm,

    y_test_lstm,

    verbose=0
)


# ============================================================
# STEP 9.2 - DISPLAY RESULTS
# ============================================================

print("\nLSTM TEST PERFORMANCE")
print("-" * 50)

print(
    f"Test Loss     : {test_loss:.4f}"
)

print(
    f"Test Accuracy : {test_accuracy:.4f}"
)

print(
    f"Test Accuracy : {test_accuracy * 100:.2f}%"
)


print("\n✓ LSTM evaluation completed")


print("\n" + "=" * 70)
print("STEP 9 COMPLETED")
print("=" * 70)

print("""
Next:
Step 10 → Generate LSTM predictions and classification metrics
""")
```

```python
# ============================================================
# THREATLENS - TASK 7
# STEP 10: LSTM CLASSIFICATION ANALYSIS
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

print("=" * 70)
print("THREATLENS - TASK 7")
print("STEP 10: LSTM CLASSIFICATION ANALYSIS")
print("=" * 70)


# ============================================================
# STEP 10.1 - GENERATE PREDICTIONS
# ============================================================

print("\nGenerating LSTM predictions...")

y_pred_lstm_probabilities = lstm_model.predict(
    X_test_lstm,
    verbose=0
)

y_pred_lstm = np.argmax(
    y_pred_lstm_probabilities,
    axis=1
)

print("✓ Predictions generated")


# ============================================================
# STEP 10.2 - CALCULATE METRICS
# ============================================================

lstm_accuracy = accuracy_score(
    y_test_lstm,
    y_pred_lstm
)

lstm_precision = precision_score(
    y_test_lstm,
    y_pred_lstm,
    average="weighted",
    zero_division=0
)

lstm_recall = recall_score(
    y_test_lstm,
    y_pred_lstm,
    average="weighted",
    zero_division=0
)

lstm_f1 = f1_score(
    y_test_lstm,
    y_pred_lstm,
    average="weighted",
    zero_division=0
)


# ============================================================
# STEP 10.3 - DISPLAY METRICS
# ============================================================

print("\n" + "=" * 70)
print("LSTM PERFORMANCE")
print("=" * 70)

print(
    f"\nAccuracy  : {lstm_accuracy:.4f}"
)

print(
    f"Precision : {lstm_precision:.4f}"
)

print(
    f"Recall    : {lstm_recall:.4f}"
)

print(
    f"F1-Score  : {lstm_f1:.4f}"
)


# ============================================================
# STEP 10.4 - CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 70)
print("LSTM CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_test_lstm,
        y_pred_lstm,
        zero_division=0
    )
)


# ============================================================
# STEP 10.5 - CONFUSION MATRIX
# ============================================================

cm_lstm = confusion_matrix(
    y_test_lstm,
    y_pred_lstm
)

print("\nConfusion Matrix:")
print(cm_lstm)


disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_lstm
)

disp.plot(
    xticks_rotation="vertical"
)

plt.title(
    "ThreatLens - LSTM Confusion Matrix"
)

plt.tight_layout()

plt.show()


print("\n" + "=" * 70)
print("STEP 10 COMPLETED")
print("=" * 70)

print("""
✓ LSTM predictions generated
✓ Accuracy calculated
✓ Precision calculated
✓ Recall calculated
✓ F1-score calculated
✓ Classification report generated
✓ Confusion matrix generated

Next:
Step 11 → LSTM error and misclassification analysis
""")
```

```python
# ============================================================
# THREATLENS - TASK 7
# STEP 11: LSTM ERROR AND MISCLASSIFICATION ANALYSIS
# ============================================================

print("=" * 70)
print("THREATLENS - TASK 7")
print("STEP 11: ERROR AND MISCLASSIFICATION ANALYSIS")
print("=" * 70)


# ============================================================
# STEP 11.1 - CALCULATE CORRECT / INCORRECT
# ============================================================

correct_predictions = np.sum(
    y_test_lstm == y_pred_lstm
)

incorrect_predictions = np.sum(
    y_test_lstm != y_pred_lstm
)

total_predictions = len(
    y_test_lstm
)

error_rate = (
    incorrect_predictions /
    total_predictions
)


print("\nLSTM ERROR SUMMARY")
print("-" * 50)

print(
    "Total test samples:",
    total_predictions
)

print(
    "Correct predictions:",
    correct_predictions
)

print(
    "Misclassified samples:",
    incorrect_predictions
)

print(
    f"Error rate: {error_rate * 100:.2f}%"
)


# ============================================================
# STEP 11.2 - MISCLASSIFICATION TABLE
# ============================================================

misclassification_data = pd.DataFrame({

    "Actual": y_test_lstm,

    "Predicted": y_pred_lstm

})


misclassification_data = (
    misclassification_data[
        misclassification_data["Actual"]
        != misclassification_data["Predicted"]
    ]
)


print("\n" + "=" * 70)
print("MISCLASSIFIED SAMPLES")
print("=" * 70)

display(
    misclassification_data.head(20)
)


# ============================================================
# STEP 11.3 - ACTUAL → PREDICTED ERROR COUNTS
# ============================================================

error_pairs = (
    misclassification_data
    .groupby(
        ["Actual", "Predicted"]
    )
    .size()
    .reset_index(
        name="Number of Errors"
    )
    .sort_values(
        "Number of Errors",
        ascending=False
    )
)


print("\n" + "=" * 70)
print("ACTUAL ATTACK → PREDICTED ATTACK")
print("=" * 70)

display(error_pairs)


# ============================================================
# STEP 11.4 - MOST COMMON ERROR
# ============================================================

if len(error_pairs) > 0:

    most_common_error = (
        error_pairs.iloc[0]
    )

    print("\nMOST COMMON MISCLASSIFICATION")
    print("-" * 50)

    print(
        "Actual class:",
        most_common_error["Actual"]
    )

    print(
        "Predicted class:",
        most_common_error["Predicted"]
    )

    print(
        "Number of errors:",
        int(
            most_common_error[
                "Number of Errors"
            ]
        )
    )

else:

    print(
        "\n✓ No misclassification found."
    )


# ============================================================
# STEP 11.5 - ERROR DISTRIBUTION BY ACTUAL CLASS
# ============================================================

if len(misclassification_data) > 0:

    error_distribution = (
        misclassification_data[
            "Actual"
        ]
        .value_counts()
        .reset_index()
    )

    error_distribution.columns = [
        "Actual Class",
        "Number of Errors"
    ]

    print("\n" + "=" * 70)
    print("ERROR DISTRIBUTION BY ACTUAL CLASS")
    print("=" * 70)

    display(error_distribution)


# ============================================================
# FINAL
# ============================================================

print("\n" + "=" * 70)
print("TASK 7 - STEP 11 COMPLETED")
print("=" * 70)

print("""
LSTM error analysis completed.

✓ Total errors identified
✓ Error rate calculated
✓ Misclassified samples identified
✓ Actual vs predicted class combinations analyzed
✓ Most common misclassification identified
✓ Error distribution by class calculated
""")
```

```python
# ============================================================
# THREATLENS - TASK 7
# STEP 12: SAVE LSTM MODEL
# ============================================================

print("=" * 70)
print("THREATLENS - TASK 7")
print("STEP 12: SAVING LSTM MODEL")
print("=" * 70)


# ============================================================
# SAVE MODEL
# ============================================================

model_path = "/content/ThreatLens_LSTM_Model.keras"

lstm_model.save(
    model_path
)

print("\n✓ LSTM model saved successfully")

print(
    "Model path:",
    model_path
)


# ============================================================
# SAVE PREDICTIONS
# ============================================================

lstm_predictions_df = pd.DataFrame({

    "Actual": y_test_lstm,

    "Predicted": y_pred_lstm
})


predictions_path = (
    "/content/ThreatLens_LSTM_Predictions.csv"
)

lstm_predictions_df.to_csv(
    predictions_path,
    index=False
)

print(
    "✓ Predictions saved:"
)

print(
    predictions_path
)


# ============================================================
# SAVE METRICS
# ============================================================

lstm_results = pd.DataFrame({

    "Model": ["LSTM"],

    "Accuracy": [lstm_accuracy],

    "Precision": [lstm_precision],

    "Recall": [lstm_recall],

    "F1-Score": [lstm_f1]
})


metrics_path = (
    "/content/ThreatLens_LSTM_Evaluation.csv"
)

lstm_results.to_csv(
    metrics_path,
    index=False
)

print(
    "✓ Evaluation metrics saved:"
)

print(
    metrics_path
)


print("\n" + "=" * 70)
print("TASK 7 COMPLETED")
print("=" * 70)

print("""
ThreatLens Task 7 completed:

✓ Data prepared for deep learning
✓ LSTM architecture designed
✓ Hyperparameters configured
✓ Dropout regularization applied
✓ Early stopping applied
✓ Learning-rate reduction applied
✓ LSTM trained
✓ Training curves generated
✓ Test performance evaluated
✓ Classification report generated
✓ Confusion matrix generated
✓ Error analysis performed
✓ Model saved
✓ Predictions saved
✓ Evaluation metrics saved
""")

print("=" * 70)
```
